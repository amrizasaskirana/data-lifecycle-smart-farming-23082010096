```python
import pandas as pd
from datetime import datetime, timedelta

# Contoh load data
df = pd.read_csv("cleaned_data.csv")  # ganti dengan path file kamu

# ==============================
# HITUNG DATA QUALITY SCORE
# ==============================

# 1. Accuracy
# Mengasumsikan "missing" adalah jumlah sel kosong di dataframe
total_cells = df.size
missing_cells = df.isnull().sum().sum()
accuracy = 1 - (missing_cells / total_cells)

# 2. Completeness
# Rasio data yang tidak null terhadap total
completeness = df.notnull().sum().sum() / total_cells

# 3. Timeliness
# Mengasumsikan ada kolom 'timestamp' dalam format datetime
if 'timestamp' in df.columns:
    df['timestamp'] = pd.to_datetime(df['timestamp'])
    last_30_days = datetime.now() - timedelta(days=30)
    recent_count = df[df['timestamp'] >= last_30_days].shape[0]
    timeliness = recent_count / len(df)
else:
    timeliness = None

# ==============================
# OUTPUT
# ==============================
print(f"Data Accuracy: {accuracy:.2%}")
print(f"Data Completeness: {completeness:.2%}")
if timeliness is not None:
    print(f"Data Timeliness (last 30 days): {timeliness:.2%}")
else:
    print("Kolom 'timestamp' tidak ditemukan, timeliness tidak dihitung.")
```
